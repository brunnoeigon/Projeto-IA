```python
# Sistema de triagem de cargas

#1. Entrada de dados

nome_carga= input("Digite o  nome da carga (texto):")
try:
  peso = float(input("Digite o peso da carga em kg (real):"))
  #2. Processamento
  if peso>=100:
    print(f"Carga pesada - rota norte")
  else:
    print (f"Carga leve - rota sul")
except ValueError:
  print ("Erro: Insira um numero valido para o peso")

#3 Estrutura de repetição: Repetir 5 vezes
for i in range (1,6):
  print (f"\n--- Triagem da mercadoria {i}---")

#4 Saida de Dados
print ('\n' + "="*30)
print ("Lote Processado com sucesso")
print ("="*30)
```
